# Regular Expressions

The main task of regular expressions (regex) is to search, match, and manipulate patterns within text. 
They allow you to efficiently find specific sequences, validate formats like emails or phone numbers, and perform complex text processing tasks with minimal code. 
Regex is a powerful tool for working with text data in a precise and flexible way.

Python has a special module called `re` that can be used to work with regular expressions.

Let's start with very basic examples.

## Regex basics

We can use regex in the similar way as we would use basic string methods to find some substring in the text:


```python
text = "Don’t explain your philosophy. Embody it."
pattern = "philosophy"

# string methods:
print("find:", text.find(pattern), "\nin:", pattern in text)
```

```python
import re

res = re.search(pattern, text)
if res is not None:
    print("res.start():", res.start())
    print("res.end():", res.end())
    print("res.span():", res.span())

print("bool(res):", bool(res))
```

This was just a basic example of what regex can do, but it’s far from showcasing their true power. 
Now, let’s dive into the key advantages and capabilities of regular expressions.

## Quantification

In regular expressions, quantifiers are special characters that specify how many times an element in the regex should be matched. 
They provide control over the number of occurrences of a character, group, or pattern in the input string. 
Here's a rundown of common quantifiers and their meanings:

| Symbol  | Meaning                                                         |
|---------:|:-----------------------------------------------------------------|
| `*`     | Matches 0 or more occurrences of the preceding element.         |
| `+`     | Matches 1 or more occurrences of the preceding element.         |
| `?`     | Matches 0 or 1 occurrence of the preceding element.             |
| `{n}`   | Matches exactly `n` occurrences of the preceding element.       |
| `{n,}`  | Matches `n` or more occurrences of the preceding element.       |
| `{n,m}` | Matches between `n` and `m` occurrences of the preceding element.|
